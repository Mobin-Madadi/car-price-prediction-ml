# Car Price Prediction — Random Forest

نوت‌بوک مرتب‌شده برای آموزش، ارزیابی و ذخیره‌سازی یک مدل **Random Forest Regressor** جهت پیش‌بینی قیمت خودرو.


## 1. Imports and configuration

```python
from pathlib import Path
import re
import warnings
import inspect

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

RANDOM_STATE = 42
DATA_PATH = Path("cars.csv")
TARGET = "price"

pd.set_option("display.max_columns", 100)
warnings.filterwarnings("ignore", category=FutureWarning)
```

## 2. Load data

```python
if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Could not find {DATA_PATH.resolve()}. "
        "Put cars.csv in the same directory as this notebook."
    )

df = pd.read_csv(DATA_PATH)
print(f"Rows: {len(df):,} | Columns: {df.shape[1]}")
display(df.head())
print("\nColumn names:")
print(df.columns.tolist())
```

## 3. Initial inspection

```python
df.info()
display(df.describe(include="all").T)
print("Missing values before cleaning:")
display(df.isna().sum().sort_values(ascending=False))
```

## 4. Data cleaning and feature engineering

- `milage` و `price` به مقادیر عددی تبدیل می‌شوند.
- قیمت‌های نامعتبر یا غیرمثبت حذف می‌شوند.
- از متن `engine`، حجم موتور و horsepower استخراج می‌شود.
- مقدارهای گمشده ویژگی‌ها را در مرحله preprocessing و فقط با داده‌های train پر می‌کنیم؛ این کار از نشت اطلاعات جلوگیری می‌کند.

```python
required_columns = {"price", "milage", "model_year"}
missing_columns = required_columns - set(df.columns)
if missing_columns:
    raise ValueError(f"Required columns are missing: {sorted(missing_columns)}")

df = df.copy()

def to_numeric_clean(series):
    # Works for values such as "$10,300", "51,000 mi.", and numeric values.
    cleaned = (
        series.astype("string")
        .str.replace(",", "", regex=False)
        .str.replace("$", "", regex=False)
        .str.replace(r"\s*mi\.?$", "", regex=True)
        .str.strip()
    )
    return pd.to_numeric(cleaned, errors="coerce")

df["price"] = to_numeric_clean(df["price"])
df["milage"] = to_numeric_clean(df["milage"])
df["model_year"] = pd.to_numeric(df["model_year"], errors="coerce")

# Remove rows where the target or essential numeric inputs are invalid.
df = df.dropna(subset=["price", "milage", "model_year"]).copy()
df = df[df["price"] > 0].copy()

# Parse engine specifications. Missing values are imputed later inside the pipeline.
if "engine" in df.columns:
    engine_text = df["engine"].astype("string")
    df["engine_size"] = pd.to_numeric(
        engine_text.str.extract(r"(\d+(?:\.\d+)?)\s*L\b", expand=False),
        errors="coerce",
    )
    df["horsepower"] = pd.to_numeric(
        engine_text.str.extract(r"(\d+(?:\.\d+)?)\s*HP\b", expand=False),
        errors="coerce",
    )

# Standardize missing-like text values and categorical columns.
missing_tokens = {"", "–", "-", "—", "nan", "None", "null"}
for col in df.select_dtypes(include=["object", "string"]).columns:
    cleaned = df[col].astype("string").str.strip()
    cleaned = cleaned.mask(cleaned.isin(missing_tokens), pd.NA)
    # Convert back to ordinary object dtype so sklearn imputers handle missing values consistently.
    df[col] = cleaned.astype(object).where(cleaned.notna(), np.nan)

print(f"Rows remaining after cleaning: {len(df):,}")
print("Target summary:")
display(df[TARGET].describe())
print("Missing values after basic cleaning:")
display(df.isna().sum().sort_values(ascending=False))
```

## 5. Quick exploratory data analysis

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

axes[0].scatter(df["milage"], df["price"], alpha=0.25)
axes[0].set(title="Price vs. Mileage", xlabel="Mileage", ylabel="Price")

axes[1].scatter(df["model_year"], df["price"], alpha=0.25)
axes[1].set(title="Price vs. Model year", xlabel="Model year", ylabel="Price")

plt.tight_layout()
plt.show()

if "brand" in df.columns:
    display(
        df.groupby("brand", dropna=False)["price"]
          .agg(["count", "median"])
          .sort_values("count", ascending=False)
          .head(15)
    )
```

## 6. Define features and make one fixed train/test split

برای جلوگیری از leakage، split فقط یک بار انجام می‌شود. تمام imputation و encoding داخل Pipeline قرار می‌گیرد و روی train fit می‌شود.

```python
candidate_numeric = ["model_year", "milage", "engine_size", "horsepower"]
candidate_categorical = [
    "brand", "model", "fuel_type", "engine", "transmission",
    "accident", "clean_title", "ext_col", "int_col"
]

numeric_features = [c for c in candidate_numeric if c in df.columns]
categorical_features = [c for c in candidate_categorical if c in df.columns]
feature_columns = numeric_features + categorical_features

if not numeric_features:
    raise ValueError("No usable numeric features were found.")

X = df[feature_columns].copy()
y = df[TARGET].copy()

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=RANDOM_STATE
)

print("Features:", feature_columns)
print("Numeric features:", numeric_features)
print("Categorical features:", categorical_features)
print(f"Train: {X_train.shape} | Test: {X_test.shape}")
```

## 7. Preprocessing pipeline

مقادیر گمشده و ویژگی‌های متنی داخل Pipeline پردازش می‌شوند؛ بنابراین همان پیش‌پردازش در زمان ارزیابی و پیش‌بینی هم استفاده خواهد شد.


```python
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
])

# سازگاری با نسخه‌های قدیمی و جدید scikit-learn
onehot_kwargs = {"handle_unknown": "ignore", "min_frequency": 10}
if "sparse_output" in inspect.signature(OneHotEncoder).parameters:
    onehot_kwargs["sparse_output"] = True
else:
    onehot_kwargs["sparse"] = True

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(**onehot_kwargs)),
])

transformers = [("numeric", numeric_transformer, numeric_features)]
if categorical_features:
    transformers.append(
        ("categorical", categorical_transformer, categorical_features)
    )

preprocessor = ColumnTransformer(
    transformers=transformers,
    remainder="drop",
)
```

## 8. Train the Random Forest model

فقط از **RandomForestRegressor** استفاده می‌کنیم. برای جلوگیری از نشت اطلاعات، پیش‌پردازش و مدل در یک Pipeline قرار دارند و فقط روی داده‌های آموزش fit می‌شوند.


```python
rf_regressor = RandomForestRegressor(
    n_estimators=300,
    max_depth=20,
    min_samples_leaf=2,
    random_state=RANDOM_STATE,
    n_jobs=-1,
)

best_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", rf_regressor),
])

best_model.fit(X_train, y_train)
test_predictions = best_model.predict(X_test)

results_df = pd.DataFrame([{
    "model": "Random Forest",
    "MAE": mean_absolute_error(y_test, test_predictions),
    "RMSE": np.sqrt(mean_squared_error(y_test, test_predictions)),
    "R2": r2_score(y_test, test_predictions),
}])

display(results_df)
```

## 9. Evaluate the Random Forest model


```python
print("Model: Random Forest")
print(f"MAE:  {mean_absolute_error(y_test, test_predictions):,.2f}")
print(f"RMSE: {np.sqrt(mean_squared_error(y_test, test_predictions)):,.2f}")
print(f"R²:   {r2_score(y_test, test_predictions):.4f}")

comparison = pd.DataFrame({
    "actual_price": y_test.to_numpy(),
    "predicted_price": test_predictions,
})
comparison["absolute_error"] = (
    comparison["actual_price"] - comparison["predicted_price"]
).abs()
display(comparison.head(10))
```

## 10. Residual analysis

```python
residuals = y_test.to_numpy() - test_predictions

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
axes[0].scatter(test_predictions, residuals, alpha=0.35)
axes[0].axhline(0, linestyle="--")
axes[0].set(
    title="Residuals vs. Predicted price",
    xlabel="Predicted price",
    ylabel="Actual - Predicted",
)

axes[1].scatter(y_test.to_numpy(), test_predictions, alpha=0.35)
low = min(y_test.min(), test_predictions.min())
high = max(y_test.max(), test_predictions.max())
axes[1].plot([low, high], [low, high], linestyle="--")
axes[1].set(
    title="Actual vs. Predicted price",
    xlabel="Actual price",
    ylabel="Predicted price",
)

plt.tight_layout()
plt.show()
```

## 11. Predict a new car

برای پیش‌بینی، باید همان ستون‌هایی را بدهیم که در زمان آموزش استفاده شده‌اند. نمونه زیر یک رکورد با اطلاعات پایه است؛ ستون‌های دیگر در صورت وجود، توسط pipeline مدیریت می‌شوند.

```python
new_car_data = {
    "model_year": 2019,
    "milage": 120_000,
    "engine_size": np.nan,
    "horsepower": np.nan,
    "brand": "Toyota",
    "model": "Unknown",
    "fuel_type": "Gasoline",
    "engine": "Unknown",
    "transmission": "Automatic",
    "accident": "Unknown",
    "clean_title": "Unknown",
    "ext_col": "Unknown",
    "int_col": "Unknown",
}

new_car = pd.DataFrame([{
    col: new_car_data.get(col, np.nan) for col in feature_columns
}])

predicted_price = best_model.predict(new_car)[0]
print(f"Estimated price: ${predicted_price:,.2f}")
```

## 12. Save the complete Random Forest pipeline

کل Pipeline شامل پیش‌پردازش و مدل ذخیره می‌شود؛ بنابراین می‌توان همین فایل را بعداً در FastAPI بارگذاری کرد.


```python
import joblib

MODEL_PATH = Path("car_price_random_forest.joblib")
joblib.dump(
    {
        "model": best_model,
        "feature_columns": feature_columns,
        "model_name": "Random Forest",
    },
    MODEL_PATH,
)

print(f"Saved model to: {MODEL_PATH.resolve()}")
```
