# Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import pylab as pl
```

## Load Dataset

```python
df = pd.read_csv('cars.csv')

df.head()
```

## Data Understanding

```python
df.head()
df.describe()
```

## Data Cleaning

```python
cdf = df[['brand' , 'model_year' , 'milage' , 'fuel_type' , 'transmission' , 'price']]

cdf.head(9)
```

## Exploratory Data Analysis

```python
plt.scatter(cdf.milage , cdf.price , color = 'blue')


plt.title("Mileage vs Price")
plt.xlabel("Mileage")
plt.ylabel("Price")


plt.show()
```
